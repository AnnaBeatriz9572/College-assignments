```python
import numpy as np
```

```python
# y = row index
# x = column index
# n = number to place in the cell
grid = [
    [0, 0, 3, 0, 2, 0, 6, 0, 0],
    [9, 0, 0, 3, 0, 5, 0, 0, 1],
    [0, 0, 1, 8, 0, 6, 4, 0, 0],
    [0, 0, 8, 1, 0, 2, 9, 0, 0],
    [7, 0, 0, 0, 0, 0, 0, 0, 8],
    [0, 0, 6, 7, 0, 8, 2, 0, 0],
    [0, 0, 2, 6, 0, 9, 5, 0, 0],
    [8, 0, 0, 2, 0, 3, 0, 0, 9],
    [0, 0, 5, 0, 1, 0, 0, 0, 2]
]
```

```python
# Check whether the number can be placed in the cell.

def possible(y, x, n):

    global grid

    for i in range(9): # Check whether the number is already in this row.
        if grid[y][i] == n:
            return False

    for i in range(9): # Check whether the number is already in this column.
        if grid[i][x] == n:
            return False

    # Check whether the number is already in this 3x3 box.
    # Calculate the starting row and column of the 3x3 box.
    # The last cell in the box is at [y0 + 2][x0 + 2].
    y0 = (y // 3) * 3 # Round down to the nearest multiple of 3.
    x0 = (x // 3) * 3

    for i in range(3):
        for j in range(3):
            if grid[y0+i][x0+j] == n:
                return False

    return True
```

```python
def check_sudoku(): # Check whether the grid follows Sudoku rules.

    global grid

    for i in range(9): # Iterate over every cell in the grid.
        for j in range(9):
            n = grid[i][j]

            if type(n) != int:
                return False

            if n < 0 or n > 9:
                return False

            if grid[i][j] != 0: # Check cells that already contain a number.
                grid[i][j] = 0
                if possible(i, j, n) == False: # Check whether this number conflicts with another cell.
                    grid[i][j] = n
                    return False
                grid[i][j] = n

                
    return True
```

```python
def check_matrix(): # Check whether the grid has 9 rows and 9 columns.
    
    global grid

    if len(grid) != 9:
        return False

    for linha in grid:
        if len(linha) != 9:
            return False

    return True
```

```python
def solve():
    global grid

    # Validate the grid before searching for solutions.
    if check_matrix() and check_sudoku():

        for y in range(9):
            for x in range(9):
                if grid[y][x] == 0:

                    for n in range(1, 10):

                        if possible(y, x, n):
                            grid[y][x] = n
                            solve()
                            grid[y][x] = 0

                    return # Return after trying all candidates for this cell.
                        # Let the previous recursive call try another candidate.
    
        print(np.matrix(grid))
    else:
        print("A matriz está incorreta")
```
