# **Parte 01 - Implementação**

Implementação dos algoritmos:
- Fatorial: recursiva e iterativa.
- Fibonacci: recursiva e iterativa.

```python
# Fatorial recursiva

def fatorial_recursiva(n):

  if n == 0 or n == 1:
    return 1

  return n * fatorial_recursiva(n-1)
```

```python
# Fatorial Iterativo
def fatorial_iterativa(n):
    fatorial = 1

    for i in range(1, n + 1):
        fatorial = fatorial * i

    return fatorial
```

```python
# Fibonacci recursiva

def fibonacci_recursiva(n):
  if n == 0:
    return 0
  elif n == 1:
    return 1
  else:
    return fibonacci_recursiva(n-1) + fibonacci_recursiva(n-2)
```

```python
# Fibonacci iterativa

def fibonacci_iterativa(n):
    if n == 0:
        return 0

    fibonacci = [0, 1]

    for i in range(2, n + 1):
        fibonacci.append(fibonacci[i - 1] + fibonacci[i - 2])

    return fibonacci[-1]
```

# **Parte 02 - Testes Experimentais**
- Testes de desempenho medindo o tempo de execução dos algoritmos.

```python
import time # para medir tempo de execução
import numpy as np # para gerar arrays
```

## **Para o fatorial**
- Para valores de entrada n = 5, 10, 50, 100, 500, 1000

```python
import sys

sys.setrecursionlimit(2000)
```

```python
lista_fatorial = [5, 10, 50, 100, 500, 1000]

fatorial_tempo_recursivo = []
fatorial_tempo_iterativo = []

for n in lista_fatorial:

    tempo_recursivo = []
    tempo_iterativo = []

    for _ in range(10):

        inicio = time.perf_counter()
        valor_recursivo = fatorial_recursiva(n)
        fim = time.perf_counter()

        tempo_recursivo.append(fim - inicio)


        inicio = time.perf_counter()
        valor_iterativo = fatorial_iterativa(n)
        fim = time.perf_counter()

        tempo_iterativo.append(fim - inicio)


    media_recursivo = np.mean(tempo_recursivo)
    media_iterativo = np.mean(tempo_iterativo)

    print("Com n =", n)
    print("Resultado recursivo:", valor_recursivo)
    print("Resultado iterativo:", valor_iterativo)
    print("Tempo médio recursivo:", media_recursivo)
    print("Tempo médio iterativo:", media_iterativo)
    print()

    fatorial_tempo_recursivo.append(media_recursivo)
    fatorial_tempo_iterativo.append(media_iterativo)
```

## Para o Fibonacci
- Para valores n = 5, 10, 20, 30, 35, 40

```python
lista_fibonacci = [5, 10, 20, 30, 35, 40]

fibonacci_tempo_recursivo = []
fibonacci_tempo_iterativo = []

for n in lista_fibonacci:

    tempo_recursivo = []
    tempo_iterativo = []

    for _ in range(10):

        inicio = time.perf_counter()
        valor_recursivo = fibonacci_recursiva(n)
        fim = time.perf_counter()

        tempo_recursivo.append(fim - inicio)


        inicio = time.perf_counter()
        valor_iterativo = fibonacci_iterativa(n)
        fim = time.perf_counter()

        tempo_iterativo.append(fim - inicio)


    media_recursivo = np.mean(tempo_recursivo)
    media_iterativo = np.mean(tempo_iterativo)

    print("Com n =", n)
    print("Resultado recursivo:", valor_recursivo)
    print("Resultado iterativo:", valor_iterativo)
    print("Tempo médio recursivo:", media_recursivo)
    print("Tempo médio iterativo:", media_iterativo)
    print()

    fibonacci_tempo_recursivo.append(media_recursivo)
    fibonacci_tempo_iterativo.append(media_iterativo)
```

# Parte 03 - Coleta de dados
- Organize os dados em uma tabela, porém como possuímos n de testes diferentes para cada algoritmo criamos nosso dataframe dessa forma:

```python
import pandas as pd
```

```python
df_fatorial = pd.DataFrame({
    "n": lista_fatorial,
    "Recursivo": fatorial_tempo_recursivo,
    "Iterativo": fatorial_tempo_iterativo
})

df_fibonacci = pd.DataFrame({
    "n": lista_fibonacci,
    "Recursivo": fibonacci_tempo_recursivo,
    "Iterativo": fibonacci_tempo_iterativo
})
```

# **Parte 04 - Gráficos Comparativos**


```python
import matplotlib.pyplot as plt
```

```python
# Gráfico de Fatorial
limite_minimo = min(df_fatorial['Iterativo'].min(), df_fatorial['Recursivo'].min())
limite_maximo = max(df_fatorial['Iterativo'].max(), df_fatorial['Recursivo'].max())

plt.plot(
    df_fatorial['n'],
    df_fatorial['Iterativo'],
    label='Iterativo',
    marker='o',
    color= 'red'
)

plt.plot(
    df_fatorial['n'],
    df_fatorial['Recursivo'],
    label='Recursivo',
    marker='o',
    color = 'blue'
)

plt.ylim(limite_minimo, limite_maximo)
plt.xlabel('n')
plt.ylabel('Tempo (s)')
plt.title('Desempenho Fatorial')
plt.grid(True)
plt.tight_layout()
plt.legend()
plt.show()
```

# Parte 05 - Análise
- Os resultados experimentais confirmam a complexidade teórica?
- Qual algoritmo foi mais eficiente? Por quê?
- Qual a principal diferença observada entre recursão e iteração?
- Por que o Fibonacci recursivo é tão lento?

Os algoritmos de fatorial e fibonacci confirmam sua complexidade teórica que são:
| Algoritmo | Complexidade assintótica |
|---|---:|
| Fatorial iterativo | \(O(n)\) |
| Fatorial recursivo | \(O(n)\) |
| Fibonacci iterativo | \(O(n)\) |
| Fibonacci recursivo | \(O(2^n)\) |


Olhando pela análise assintótica podemos perceber pelos cálculos que o algoritmo de fatorial se torna mais eficiente pelo fato de sua complexidade ser O(n) em ambos os casos, tanto iterativo quanto recursivo, ao contrário do Fibonacci que apresentou o algoritmo recursivo O(2^n). Enquanto na análise experimental podemos ver o mesmo pensamento analítico por causa da representação em gráfico que apresenta o máximo obtido nos algoritmos de fibonacci chegando a 20 segundos na recursiva, e nos outros o máximo chegava a 10^(-4) segundos.

Durante o processo podemos ver algumas das principais diferenças entre recursão e iteração. Além do fato de um utilizar de chamadas de funções de forma sucessiva e o outro utilizar de laços de repetição, observamos que nos algoritmos de iteração eles possuem os menores tempos de execução.

Portanto, o Fibonacci recursivo é lento pelo fato de repetir muitos cálculos já realizados, gerando uma grande quantidade de chamadas recursivas e crescimento exponencial do tempo de execução.
